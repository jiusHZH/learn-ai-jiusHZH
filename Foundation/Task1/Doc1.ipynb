## 1. 基础容器：List（列表）与 Dict（字典）

- List 是有序集合，可以存储多个元素，支持追加、修改、切片等操作。
- Dict 是键值对集合，适合按名称访问数据，查找速度快。
- 在 Python 中，这两种结构是最常见的基础容器之一。

```python
# 基础容器：List 与 Dict
# 列表适合存放有序、可变的数据；字典适合按键值对存储数据。

# List：常用操作
nums = [1, 2, 3, 4]
nums.append(5)
nums[0] = 10
print('List:', nums)
print('第二个元素:', nums[1])
print('切片:', nums[:3])

# Dict：按 key 访问
student = {'name': 'Alice', 'age': 20, 'score': 95}
student['class'] = 'Python'
print('Dict:', student)
print('姓名:', student['name'])
print('keys:', list(student.keys()))
print('items:', list(student.items()))
```

## 2. 函数：Lambda 匿名函数与 Decorator 装饰器

- Lambda 是简洁的匿名函数，适合小规模逻辑。
- Decorator 可以在不修改原函数代码的情况下，给函数增加额外功能。
- 这是 Python 中非常常见的高级语法。

```python
# Lambda 匿名函数：适合一行简单逻辑
square = lambda x: x * x
print('平方:', square(6))

# Decorator：给函数增加额外功能

def log_time(func):
    def wrapper(*args, **kwargs):
        print('调用函数前：开始执行')
        result = func(*args, **kwargs)
        print('调用函数后：执行结束')
        return result
    return wrapper

@log_time
def greet(name):
    print(f'Hello, {name}!')
    return 'done'

print(greet('Python'))
```

## 3. 面向对象：Class 与 Magic Methods

面向对象编程（OOP）把数据和行为封装成对象。类负责定义模板，对象负责保存具体值。

- `__init__` 是构造方法，用于初始化对象。
- `__str__` 和 `__repr__` 是常见魔法方法，用于对象的字符串表示。
- OOP 的核心思想包括封装、继承和多态。

```python
# 面向对象：类与魔法方法
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        return f'{self.name} 在叫：汪汪！'

    def __str__(self):
        return f'Dog(name={self.name}, age={self.age})'

    def __len__(self):
        return self.age

my_dog = Dog('小白', 3)
print(my_dog.bark())
print(str(my_dog))
print(len(my_dog))

# OOP 思想：封装、继承、多态
class Animal:
    def speak(self):
        return '发声'

class Cat(Animal):
    def speak(self):
        return '喵喵'

animals = [Cat()]
for a in animals:
    print(a.speak())
```

## 4. 文本处理：re 正则表达式

正则表达式（regular expression）是用来匹配文本模式的工具。

- `re.findall()`：查找所有匹配项
- `re.search()`：搜索第一个匹配
- `re.sub()`：替换匹配内容
- 常见模式：`\d` 表示数字、`\w` 表示字母/数字/下划线、`[A-Za-z]+` 表示英文单词

下面是一个简单示例。

```python
import re

text = '联系电话：138-1234-5678，邮箱：alice@example.com，编号：A123'

# 查找所有数字
numbers = re.findall(r'\d+', text)
print('数字:', numbers)

# 搜索邮箱
match = re.search(r'[A-Za-z0-9_.+-]+@[A-Za-z0-9-]+\.[A-Za-z]{2,}', text)
if match:
    print('邮箱:', match.group(0))

# 替换电话号中的 '-' 为空格
masked = re.sub(r'-', ' ', '138-1234-5678')
print('替换后:', masked)
```

## 5. 代码格式：列表推导式与 Type Hint

- 列表推导式是一种简洁的创建列表方式。
- Type Hint 表示类型注释，可以让代码更清晰，也方便 IDE 和静态检查工具分析。

这类写法能让 Python 代码更简洁、更易维护。

```python
# 列表推导式：快速生成列表
squares = [x * x for x in range(1, 7)]
print('平方列表:', squares)

# Type Hint：类型注释
from typing import List, Dict

def average(nums: List[float]) -> float:
    return sum(nums) / len(nums)

scores: List[float] = [90.0, 85.5, 88.0]
print('平均分:', average(scores))

user: Dict[str, str] = {'name': 'Bob', 'role': 'student'}
print('用户信息:', user)
```

## 6. 进阶技巧：Generator 生成器（yield）

生成器从迭代器进一步演化而来，使用 `yield` 关键字返回值，并保持函数的状态。它适合处理大规模数据流，因为它是惰性计算的。

- `yield` 会暂停函数执行
- 下次调用 `next()` 时会继续执行
- 适合迭代大量数据，节省内存

下面演示一个简单生成器。

```python
# Generator 生成器函数：用 yield 逐步产出值

def countdown(n):
    while n > 0:
        yield n
        n -= 1 #能用 for 的话，这里就不需要控制

for value in countdown(5):
    print('当前值:', value)

# 生成器也可以用 next() 手动迭代
nums = countdown(3)
print(next(nums))
print(next(nums))
print(next(nums))

#还有一种方式是生成器表达式，与列表推导式类似，但使用圆括号而不是方括号
gen_expr = (x * x for x in range(1, 6))
next(gen_expr)  # 第一个值，后续直接从 4 开始
for value in gen_expr:
    print('生成器表达式值:', value)
```
